## ✏️ Sección 3: Operaciones matemáticas

### 🟢 Ejercicio 3.1 — Calculadora de aula

Tienes un salón con estas variables. Calcula e imprime los resultados:

```python
total_estudiantes = 32
mesas = 8
lapices_por_caja = 12
cajas = 5
```

Calcula:
1. ¿Cuántos estudiantes hay por mesa?
2. ¿Cuántos lápices hay en total?
3. ¿Alcanzan los lápices para todos? (lápices totales - estudiantes)
4. ¿Cuántos lápices sobran o faltan?

````python
# ✍️ Escribe tu solución aquí:
total_estudiantes = 32
mesas = 8
lapices_por_caja = 12
cajas = 5
````

````python
# 👩‍🏫 SOLUCIÓN
total_estudiantes = 32
mesas = 8
lapices_por_caja = 12
cajas = 5

por_mesa = total_estudiantes / mesas
total_lapices = lapices_por_caja * cajas
diferencia = total_lapices - total_estudiantes

print(f"Estudiantes por mesa: {por_mesa}")
print(f"Lápices en total: {total_lapices}")

if diferencia >= 0:
    print(f"Sobran {diferencia} lápices ✅")
else:
    print(f"Faltan {abs(diferencia)} lápices ❌")
````

### 🟡 Ejercicio 3.2 — Conversor de temperatura

La fórmula para convertir Celsius a Fahrenheit es:  
`F = (C × 9/5) + 32`

La fórmula para convertir Celsius a Kelvin es:  
`K = C + 273.15`

Dada una temperatura de `25°C`, calcula e imprime las equivalencias en Fahrenheit y Kelvin.

**Salida esperada:**
```
25°C equivale a:
  → 77.0°F
  → 298.15 K
```

````python
# ✍️ Escribe tu solución aquí:
celsius = 25
````

````python
# 👩‍🏫 SOLUCIÓN
celsius = 25
fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

print(f"{celsius}°C equivale a:")
print(f"  → {fahrenheit}°F")
print(f"  → {kelvin} K")
````

## ✏️ Sección 4: `input()` — Entrada del usuario

### 🟡 Ejercicio 4.1 — Calculadora de IMC

El Índice de Masa Corporal (IMC) se calcula así:  
`IMC = peso / (altura²)`

Pide al usuario:
1. Su nombre
2. Su peso en kg (número decimal)
3. Su altura en metros (número decimal)

Luego calcula e imprime el IMC con 2 decimales.

**Salida esperada:**
```
¿Cuál es tu nombre? Ana
¿Cuánto pesas (kg)? 65
¿Cuánto mides (m)? 1.65

Ana, tu IMC es: 23.88
```

````python
# ✍️ Escribe tu solución aquí:
````

````python
# 👩‍🏫 SOLUCIÓN
nombre = input("¿Cuál es tu nombre? ")
peso = float(input("¿Cuánto pesas (kg)? "))
altura = float(input("¿Cuánto mides (m)? "))

imc = peso / (altura ** 2)
print(f"\n{nombre}, tu IMC es: {imc:.2f}")
````

## ✏️ Sección 5: Condicionales `if / elif / else`

### 🟢 Ejercicio 5.1 — ¿Número par o impar?

Escribe un programa que pida un número al usuario y le diga si es **par** o **impar**.

_Pista: un número es par si el residuo de dividirlo entre 2 es 0 (`numero % 2 == 0`)_

**Ejemplos:**
```
Ingresa un número: 7   →  7 es impar
Ingresa un número: 12  →  12 es par
```

````python
# ✍️ Escribe tu solución aquí:
````

````python
# 👩‍🏫 SOLUCIÓN
numero = int(input("Ingresa un número: "))

if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
````

### 🟡 Ejercicio 5.2 — Clasificador de IMC

Mejora el ejercicio del IMC (4.1) agregando una clasificación según la OMS:

| IMC | Clasificación |
|-----|---------------|
| Menor de 18.5 | Bajo peso |
| 18.5 a 24.9 | Peso normal |
| 25 a 29.9 | Sobrepeso |
| 30 o más | Obesidad |

**Salida esperada:**
```
Tu IMC es: 23.88 → Peso normal ✅
```

````python
# ✍️ Escribe tu solución aquí:
# (puedes partir del ejercicio 4.1)
````

````python
# 👩‍🏫 SOLUCIÓN
nombre = input("¿Cuál es tu nombre? ")
peso = float(input("¿Cuánto pesas (kg)? "))
altura = float(input("¿Cuánto mides (m)? "))

imc = peso / (altura ** 2)

if imc < 18.5:
    clasificacion = "Bajo peso ⚠️"
elif imc < 25:
    clasificacion = "Peso normal ✅"
elif imc < 30:
    clasificacion = "Sobrepeso ⚠️"
else:
    clasificacion = "Obesidad ❌"

print(f"\n{nombre}, tu IMC es: {imc:.2f} → {clasificacion}")
````

### 🔴 Reto 1 — Registro de estudiante

Crea un programa que:

1. Pida el **nombre** del estudiante
2. Pida **3 notas** (entre 0 y 10)
3. Calcule el **promedio**
4. Determine si **aprobó o reprobó** (mínimo para aprobar: 6.0)
5. Muestre un **informe completo**

**Salida esperada:**
```
============================
    INFORME DE ESTUDIANTE   
============================
Nombre: Lucía Morales
Notas: 8.5, 6.0, 7.5
```

### 🔴 Reto 2 — Calculadora de viaje

Crea un programa que ayude a planear un viaje:

1. Pide el **destino** (texto)
2. Pide la **distancia en km** (entero)
3. Pide el **consumo del vehículo en km por litro** (decimal)
4. Pide el **precio del combustible por litro** en pesos (decimal)
5. Calcula:
   - Litros necesarios
   - Costo total del combustible
   - Si la distancia es mayor a 500 km, recomienda hacer una parada
6. Muestra un **resumen del viaje**

````python
# ✍️ Escribe tu solución aquí:
````

````python
# 👩‍🏫 SOLUCIÓN
destino = input("¿A dónde vas? ")
distancia = int(input("¿Cuántos km hay hasta el destino? "))
consumo = float(input("¿Cuántos km hace tu vehículo por litro? "))
precio_litro = float(input("¿Cuánto cuesta el litro de combustible (pesos)? "))

litros_necesarios = distancia / consumo
costo_total = litros_necesarios * precio_litro

print(f"\n🚗 RESUMEN DEL VIAJE A {destino.upper()}")
print(f"Distancia: {distancia} km")
print(f"Combustible necesario: {litros_necesarios:.1f} litros")
print(f"Costo total: ${costo_total:,.0f} pesos")

if distancia > 500:
    print("\n⚠️ Viaje largo: se recomienda hacer al menos una parada para descansar.")
else:
    print("\n✅ Viaje corto: puedes ir sin paradas intermedias.")
````
