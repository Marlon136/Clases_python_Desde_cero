## 👋 2. Tu primer programa: `print()`

## 📦 3. Variables y Tipos de Datos

Una **variable** es como una caja con nombre donde guardamos información. Piensa en ellas como etiquetas en cajitas de almacenamiento.

```
nombre = "Ana"   →  📦 nombre  contiene  "Ana"
edad   = 35      →  📦 edad    contiene  35
```

### Tipos de datos principales en Python:

| Tipo | Nombre técnico | Ejemplo |
|------|---------------|----------|
| Texto | `str` (string) | `"Hola"`, `'Python'` |
| Número entero | `int` | `10`, `-3`, `100` |
| Número decimal | `float` | `3.14`, `9.99` |
| Verdadero/Falso | `bool` | `True`, `False` |

````python
# Creando variables — el símbolo = significa 'guardar en'
nombre = "Ana"
apellido = "García"
edad = 35
altura = 1.65
es_profesora = True

# Mostramos los valores
print(nombre)
print(edad)
print(es_profesora)
````

````python
# La función type() nos dice qué tipo de dato es una variable
print(type(nombre))       # str = texto
print(type(edad))         # int = número entero
print(type(altura))       # float = número decimal
print(type(es_profesora)) # bool = verdadero/falso
````

````python
# f-strings: la forma moderna de combinar texto y variables
# Ponemos una f antes de las comillas y las variables van entre { }

print(f"Me llamo {nombre} {apellido}")
print(f"Tengo {edad} años y mido {altura} metros")
print(f"¿Es profesora? {es_profesora}")
````

````python
# Las variables pueden cambiar su valor
puntuacion = 0
print(f"Puntuación inicial: {puntuacion}")

puntuacion = 10
print(f"Después de responder bien: {puntuacion}")

puntuacion = puntuacion + 5   # le sumamos 5 a lo que ya tenía
print(f"Después de otro acierto: {puntuacion}")
````

## ➕ 4. Operaciones Matemáticas

Python puede funcionar como una calculadora muy poderosa.

| Operación | Símbolo | Ejemplo |
|-----------|---------|----------|
| Suma | `+` | `5 + 3` → `8` |
| Resta | `-` | `10 - 4` → `6` |
| Multiplicación | `*` | `6 * 7` → `42` |
| División | `/` | `15 / 4` → `3.75` |
| División entera | `//` | `15 // 4` → `3` |
| Módulo (residuo) | `%` | `15 % 4` → `3` |
| Potencia | `**` | `2 ** 8` → `256` |

````python
# Operaciones básicas
print(10 + 5)    # Suma
print(20 - 8)    # Resta
print(4 * 7)     # Multiplicación
print(18 / 4)    # División (resultado decimal)
````

````python
# Operaciones más interesantes
print(17 // 5)   # División entera: ¿cuántas veces cabe 5 en 17?
print(17 % 5)    # Módulo: ¿qué sobra de dividir 17 entre 5?
print(2 ** 10)   # Potencia: 2 elevado a la 10
````

````python
# Usando variables en los cálculos — así es más útil y legible
precio_unitario = 15000
cantidad = 30
descuento = 0.10  # 10%

total = precio_unitario * cantidad
ahorro = total * descuento
total_final = total - ahorro

print(f"Total sin descuento: ${total:,}")
print(f"Descuento (10%):    ${ahorro:,}")
print(f"Total a pagar:      ${total_final:,}")
````

## ⌨️ 5. Entrada del Usuario: `input()`

La función `input()` **pausa el programa y espera a que el usuario escriba algo**. Todo lo que el usuario escribe llega como texto (`str`), así que si necesitamos un número, debemos convertirlo.

| Conversión | Función | Uso |
|-----------|---------|-----|
| Texto → Entero | `int()` | `int("25")` → `25` |
| Texto → Decimal | `float()` | `float("3.5")` → `3.5` |
| Número → Texto | `str()` | `str(100)` → `"100"` |

````python
# El programa pregunta y espera respuesta
nombre_usuario = input("¿Cómo te llamas? ")
print(f"¡Hola, {nombre_usuario}! Bienvenida a Python 🐍")
````

````python
# Cuando necesitamos un número, convertimos con int() o float()
edad_texto = input("¿Cuántos años tienes? ")
edad_numero = int(edad_texto)         # convertimos texto a número

anios_hasta_jubilacion = 65 - edad_numero
print(f"Te faltan {anios_hasta_jubilacion} años para jubilarte.")
````

````python
# Versión más elegante: convertimos directamente en la misma línea
nota1 = float(input("Ingresa la primera nota (0-10): "))
nota2 = float(input("Ingresa la segunda nota (0-10): "))
nota3 = float(input("Ingresa la tercera nota (0-10): "))

promedio = (nota1 + nota2 + nota3) / 3
print(f"\nTu promedio es: {promedio:.1f}")
````

## 🔀 6. Condicionales: `if / elif / else`

Los condicionales le permiten al programa **tomar decisiones**. Son la base de la lógica de programación.

```
si (condición es verdadera):
    hacer esto
si no, si (otra condición):
    hacer esto otro
si no:
    hacer esto por defecto
```

### Operadores de comparación:

| Operador | Significado | Ejemplo |
|----------|------------|----------|
| `==` | igual a | `5 == 5` → `True` |
| `!=` | diferente de | `5 != 3` → `True` |
| `>` | mayor que | `8 > 3` → `True` |
| `<` | menor que | `2 < 9` → `True` |
| `>=` | mayor o igual | `5 >= 5` → `True` |
| `<=` | menor o igual | `4 <= 7` → `True` |

> ⚠️ **¡Ojo!** `=` es para **asignar** un valor. `==` es para **comparar**.

````python
# Ejemplo sencillo: verificar si es mayor de edad
edad = 20

if edad >= 18:
    print("Es mayor de edad ✅")
else:
    print("Es menor de edad ❌")
````

````python
# Ejemplo con elif: calificar una nota
nota = 7.5

if nota >= 9:
    calificacion = "Excelente 🌟"
elif nota >= 7:
    calificacion = "Bueno 👍"
elif nota >= 5:
    calificacion = "Suficiente ⚠️"
else:
    calificacion = "Insuficiente ❌"

print(f"Nota: {nota} → {calificacion}")
````

````python
# Combinando input() con condicionales: un programa real
temperatura = float(input("¿Cuál es la temperatura de hoy (°C)? "))

if temperatura >= 30:
    print("🌡️ Hace mucho calor. ¡Hidrátate bien!")
elif temperatura >= 20:
    print("☀️ Temperatura agradable. ¡Buen día!")
elif temperatura >= 10:
    print("🧥 Está fresco. Lleva una chaqueta.")
else:
    print("🥶 Hace frío. ¡Abrígate!")
````

````python
# Operadores lógicos: and, or, not
# 'and' = las DOS condiciones deben ser verdaderas
# 'or'  = al menos UNA condición debe ser verdadera
# 'not' = invierte la condición

tiene_cédula = True
tiene_pase = False

if tiene_cédula and tiene_pase:
    print("Puede ingresar ✅")
elif tiene_cédula or tiene_pase:
    print("Puede ingresar con supervisión ⚠️")
else:
    print("No puede ingresar ❌")
````
