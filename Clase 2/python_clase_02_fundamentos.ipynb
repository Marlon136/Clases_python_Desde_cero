## 📋 1. Listas — Guardar muchos datos juntos

Hasta ahora guardábamos **un dato por variable**. Pero en un salón tienes 30 estudiantes. ¿Crearías 30 variables?

```python
# ❌ Así NO lo harías
estudiante1 = "Lucía"
estudiante2 = "Tomás"
estudiante3 = "Valeria"
# ...hasta 30
```

Una **lista** agrupa muchos valores en una sola variable, en orden:

```
nombres = ["Lucía", "Tomás", "Valeria", ...]
              0         1        2
             👆 los índices empiezan en 0
```

````python
# Creando una lista de estudiantes del salón
estudiantes = ["Lucía", "Tomás", "Valeria", "Andrés", "Camila"]

print(estudiantes)              # toda la lista
print(estudiantes[0])           # primer elemento (índice 0)
print(estudiantes[2])           # tercer elemento (índice 2)
print(estudiantes[-1])          # último elemento (índice -1)
print(f"Total: {len(estudiantes)} estudiantes")  # len() cuenta los elementos
````

````python
# Las listas pueden tener cualquier tipo de dato
notas = [8.5, 6.0, 9.5, 5.0, 7.5]

print(notas)
print(f"Nota más alta: {max(notas)}")   # max() da el valor mayor
print(f"Nota más baja: {min(notas)}")   # min() da el valor menor
print(f"Suma de notas: {sum(notas)}")   # sum() suma todos
print(f"Promedio: {sum(notas)/len(notas):.2f}")
````

````python
# Las listas se pueden modificar: agregar, cambiar y eliminar elementos
materias = ["Matemáticas", "Lenguaje", "Ciencias"]
print(f"Lista inicial: {materias}")

materias.append("Tecnología")        # agregar al final
print(f"Después de append: {materias}")

materias[1] = "Español"              # cambiar un elemento
print(f"Después de cambiar [1]: {materias}")

materias.remove("Ciencias")          # eliminar por valor
print(f"Después de remove: {materias}")

materias.sort()                      # ordenar alfabéticamente
print(f"Ordenada: {materias}")
````

````python
# Slicing: obtener partes de una lista [inicio:fin]
# (el índice 'fin' no se incluye)
top_estudiantes = ["Ana", "Beto", "Carlos", "Diana", "Elena", "Felipe"]

print(top_estudiantes[0:3])    # primeros 3
print(top_estudiantes[2:5])    # del índice 2 al 4
print(top_estudiantes[:3])     # desde el inicio hasta el 3
print(top_estudiantes[3:])     # desde el 3 hasta el final
````

## 📌 2. Tuplas — Listas que no cambian

Una **tupla** es como una lista, pero **inmutable**: una vez creada, no se puede modificar. Se usa para datos que no deben cambiar.

```
Lista  →  [  ]  →  puede cambiar
Tupla  →  (  )  →  NO puede cambiar
```

**Ejemplos de uso:** días de la semana, meses del año, coordenadas, niveles de grado.

````python
# Tuplas: datos fijos del colegio
dias_clase = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes")
grados = ("6°", "7°", "8°", "9°", "10°", "11°")
rango_notas = (0.0, 10.0)   # mínimo y máximo posible

print(dias_clase)
print(f"Hay clase el: {dias_clase[0]}")
print(f"Grados del colegio: {grados}")
print(f"Las notas van de {rango_notas[0]} a {rango_notas[1]}")
````

````python
# Las tuplas NO se pueden modificar — esto generará un error (¡intencionalmente!)
# Descomenta la siguiente línea para ver el error:

# dias_clase[0] = "Domingo"  # ← TypeError: 'tuple' object does not support item assignment

print("Las tuplas protegen datos que no deben cambiar 🔒")
print(f"Tipo de dato: {type(dias_clase)}")
````

## 🔄 3. Bucle `for` — Repetir acciones

Un bucle `for` recorre una lista (u otra colección) y ejecuta un bloque de código **por cada elemento**.

```python
for elemento in coleccion:
    # esto se repite para cada elemento
```

Piénsalo así: *"Para cada estudiante en la lista de estudiantes, haz esto..."*

````python
# Pasar lista con for: saludar a cada estudiante
estudiantes = ["Lucía", "Tomás", "Valeria", "Andrés", "Camila"]

for nombre in estudiantes:
    print(f"Buenos días, {nombre} 👋")
````

````python
# range(): generar una secuencia de números
# range(5)       → 0, 1, 2, 3, 4
# range(1, 6)    → 1, 2, 3, 4, 5
# range(1, 11, 2)→ 1, 3, 5, 7, 9  (de 2 en 2)

print("Tabla del 7:")
for i in range(1, 11):
    resultado = 7 * i
    print(f"  7 × {i} = {resultado}")
````

````python
# enumerate(): recorrer con índice y valor al mismo tiempo
# Muy útil para mostrar listas numeradas

materias = ["Matemáticas", "Español", "Ciencias", "Tecnología", "Educación Física"]

print("📚 Materias del semestre:")
for numero, materia in enumerate(materias, start=1):
    print(f"  {numero}. {materia}")
````

````python
# Calcular estadísticas del salón con for
notas = [8.5, 6.0, 9.5, 5.0, 7.5, 8.0, 4.5, 9.0, 7.0, 6.5]

aprobados = 0
reprobados = 0

for nota in notas:
    if nota >= 6.0:
        aprobados += 1       # equivale a: aprobados = aprobados + 1
    else:
        reprobados += 1

total = len(notas)
promedio = sum(notas) / total

print(f"📊 Resultados del grupo:")
print(f"   Total estudiantes: {total}")
print(f"   Aprobados:  {aprobados} ({aprobados/total*100:.0f}%)")
print(f"   Reprobados: {reprobados} ({reprobados/total*100:.0f}%)")
print(f"   Promedio del grupo: {promedio:.2f}")
````

````python
# zip(): recorrer dos listas al mismo tiempo
# Ideal para relacionar nombres con sus notas

nombres  = ["Lucía", "Tomás", "Valeria", "Andrés", "Camila"]
notas    = [8.5,     6.0,     9.5,       5.0,      7.5]

print("📝 INFORME DE NOTAS")
print("-" * 30)
for nombre, nota in zip(nombres, notas):
    estado = "✅" if nota >= 6.0 else "❌"
    print(f"{estado} {nombre:<10} → {nota}")
````

## 🔁 4. Bucle `while` — Repetir con condición

El bucle `while` repite un bloque **mientras una condición sea verdadera**. A diferencia del `for`, no sabe cuántas veces va a repetirse: sigue hasta que la condición se vuelva falsa.

```python
while (condición es True):
    # hacer esto
    # en algún momento la condición debe volverse False
    # o el programa nunca terminará (bucle infinito ⚠️)
```

**Regla de oro:** siempre asegúrate de que algo dentro del `while` haga que la condición eventualmente sea `False`.

````python
# Cuenta regresiva para el recreo
minutos = 5

while minutos > 0:
    print(f"⏱️ Faltan {minutos} minuto(s) para el recreo...")
    minutos -= 1    # equivale a: minutos = minutos - 1

print("🎉 ¡Recreo!")
````

````python
# Ingresar notas hasta que el usuario decida parar
# (simulamos una entrada manual con una lista predefinida para no usar input() en el demo)

notas_ingresadas = []
entradas = ["8.5", "7.0", "9.0", "5.5", "listo"]  # simulación de lo que escribiría el usuario
indice = 0

print("Ingresa notas una por una. Escribe 'listo' para terminar.")

while True:                             # bucle infinito que nosotros rompemos
    entrada = entradas[indice]          # en la clase real: entrada = input("Nota: ")
    indice += 1
    print(f"  > Nota ingresada: {entrada}")

    if entrada.lower() == "listo":
        break                           # break sale del bucle inmediatamente

    notas_ingresadas.append(float(entrada))

print(f"\nNotas registradas: {notas_ingresadas}")
print(f"Promedio: {sum(notas_ingresadas)/len(notas_ingresadas):.2f}")
````

````python
# 🆚 ¿Cuándo usar for y cuándo while?

# FOR → cuando sabes CUÁNTAS veces repetir
print("=== FOR: recorrer lista de estudiantes ===")
estudiantes = ["Ana", "Luis", "Marta"]
for e in estudiantes:
    print(f"  Pasando lista a: {e}")

print()

# WHILE → cuando no sabes cuántas veces, dependes de una condición
print("=== WHILE: esperar que un estudiante apruebe ===")
intentos = 0
nota = 4.0
while nota < 6.0:
    intentos += 1
    nota += 1.5
    print(f"  Intento {intentos}: nota = {nota}")
print(f"  ✅ ¡Aprobó en el intento {intentos}!")
````

## 🗂️ 5. Diccionarios — Datos con etiqueta

Un **diccionario** guarda pares de **clave: valor**. Es como una ficha de estudiante donde cada campo tiene un nombre.

```
Lista       →  ["Lucía", 8.5, "8°"]          # accedes por posición [0], [1]...
Diccionario →  {"nombre": "Lucía",            # accedes por nombre ["nombre"]
                "nota": 8.5,
                "grado": "8°"}
```

Los diccionarios hacen el código **mucho más legible** cuando los datos tienen significado.

````python
# Ficha de un estudiante como diccionario
estudiante = {
    "nombre": "Lucía Martínez",
    "grado": "8°",
    "edad": 14,
    "promedio": 8.5,
    "aprobado": True
}

# Acceder a valores por su clave
print(estudiante["nombre"])
print(estudiante["grado"])
print(f"Promedio: {estudiante['promedio']}")
````

````python
# Modificar y agregar claves
estudiante["promedio"] = 9.0          # cambiar valor existente
estudiante["email"] = "lucia@edu.co"  # agregar nueva clave

print(estudiante)
````

````python
# Métodos útiles de diccionarios
ficha = {
    "nombre": "Tomás Ruiz",
    "materia": "Tecnología",
    "nota": 7.5
}

print(ficha.keys())    # todas las claves
print(ficha.values())  # todos los valores
print(ficha.items())   # pares clave-valor

# .get() es más seguro que [] porque no da error si la clave no existe
print(ficha.get("nota", "Sin nota"))       # existe → devuelve 7.5
print(ficha.get("telefono", "Sin dato"))   # no existe → devuelve el valor por defecto
````

````python
# Recorrer un diccionario con for
ficha = {
    "Nombre": "Valeria López",
    "Grado": "9°",
    "Edad": 15,
    "Ciudad": "Medellín",
    "Promedio": 9.2
}

print("📋 FICHA DEL ESTUDIANTE")
print("=" * 25)
for clave, valor in ficha.items():
    print(f"  {clave}: {valor}")
````

````python
# El poder real: una lista de diccionarios = base de datos simple del salón
salon = [
    {"nombre": "Lucía",   "nota": 8.5},
    {"nombre": "Tomás",   "nota": 5.0},
    {"nombre": "Valeria", "nota": 9.5},
    {"nombre": "Andrés",  "nota": 6.0},
    {"nombre": "Camila",  "nota": 4.5},
]

print("📊 BOLETÍN DEL GRUPO")
print(f"{'Nombre':<12} {'Nota':>6} {'Estado':>10}")
print("-" * 32)

for est in salon:
    nombre = est["nombre"]
    nota   = est["nota"]
    estado = "Aprobado ✅" if nota >= 6.0 else "Reprobado ❌"
    print(f"{nombre:<12} {nota:>6} {estado:>10}")

promedio_grupo = sum(e["nota"] for e in salon) / len(salon)
print("-" * 32)
print(f"Promedio del grupo: {promedio_grupo:.2f}")
````

## 🎯 Resumen de la Clase

| Concepto | ¿Para qué sirve? | Sintaxis clave |
|----------|-----------------|----------------|
| `list` | Guardar muchos valores en orden | `notas = [8.5, 7.0, 9.0]` |
| `tuple` | Datos fijos que no cambian | `dias = ("Lunes", "Martes")` |
| `for` | Repetir para cada elemento | `for n in notas:` |
| `range()` | Generar secuencias de números | `range(1, 11)` |
| `enumerate()` | Recorrer con índice y valor | `for i, v in enumerate(lista):` |
| `zip()` | Recorrer dos listas a la vez | `for n, nota in zip(nombres, notas):` |
| `while` | Repetir mientras algo sea True | `while intentos < 3:` |
| `dict` | Datos con nombre (clave:valor) | `{"nombre": "Ana", "nota": 9}` |
