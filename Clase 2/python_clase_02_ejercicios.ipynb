### 🟡 Ejercicio 2.3 — Boletín de notas con `zip()`

Tienes dos listas paralelas. Usa `zip()` y un `for` para imprimir el boletín, incluyendo el estado (aprobado/reprobado) y el promedio al final:

```python
nombres = ["Ana", "Beto", "Clara", "David", "Emma"]
notas   = [9.0,   5.5,    7.0,     4.0,     8.5]
```

**Salida esperada:**
```
NOMBRE       NOTA   ESTADO
------------------------------
✅ Ana         9.0  Aprobado
❌ Beto        5.5  Reprobado
...
------------------------------
Promedio del grupo: 6.80
```

````python
# ✍️ Escribe tu solución aquí:
nombres = ["Ana", "Beto", "Clara", "David", "Emma"]
notas   = [9.0,   5.5,    7.0,     4.0,     8.5]
````

````python
# 👩‍🏫 SOLUCIÓN
nombres = ["Ana", "Beto", "Clara", "David", "Emma"]
notas   = [9.0,   5.5,    7.0,     4.0,     8.5]

print(f"{'NOMBRE':<12} {'NOTA':>5}   ESTADO")
print("-" * 30)

for nombre, nota in zip(nombres, notas):
    if nota >= 6.0:
        icono, estado = "✅", "Aprobado"
    else:
        icono, estado = "❌", "Reprobado"
    print(f"{icono} {nombre:<10} {nota:>5}  {estado}")

print("-" * 30)
print(f"Promedio del grupo: {sum(notas)/len(notas):.2f}")
````

## ✏️ Sección 3: Bucle `while`

### 🟢 Ejercicio 3.1 — Contador de intentos

Un estudiante tiene máximo **3 intentos** para adivinar la materia secreta. Usa `while` para:
1. Contar los intentos
2. Salir si adivina (usa `break`)
3. Salir si se acaban los intentos
4. Dar un mensaje diferente según el resultado

```python
materia_secreta = "tecnología"
intentos_simulados = ["matemáticas", "español", "tecnología"]  # simula lo que escribe
```

````python
# ✍️ Escribe tu solución aquí:
materia_secreta = "tecnología"
intentos_simulados = ["matemáticas", "español", "tecnología"]
````

````python
# 👩‍🏫 SOLUCIÓN
materia_secreta = "tecnología"
intentos_simulados = ["matemáticas", "español", "tecnología"]
max_intentos = 3
intento = 0
adivino = False

while intento < max_intentos:
    respuesta = intentos_simulados[intento]   # en clase real: input("Adivina la materia: ")
    intento += 1
    print(f"Intento {intento}: '{respuesta}'")

    if respuesta.lower() == materia_secreta:
        adivino = True
        break
    else:
        restantes = max_intentos - intento
        if restantes > 0:
            print(f"  Incorrecto. Te quedan {restantes} intento(s).")

if adivino:
    print(f"\n🎉 ¡Correcto! Era '{materia_secreta}'. Lo lograste en {intento} intento(s).")
else:
    print(f"\n😞 Se acabaron los intentos. Era '{materia_secreta}'.")
````

### 🟡 Ejercicio 3.2 — Acumulador de notas

Simula un programa que acumula notas hasta que el usuario escriba `"fin"`. Al terminar, muestra el promedio y cuántas notas se ingresaron.

```python
# Simula las entradas del usuario:
entradas = ["8.0", "6.5", "9.0", "7.5", "fin"]
```

**Salida esperada:**
```
> Nota ingresada: 8.0
> Nota ingresada: 6.5
> Nota ingresada: 9.0
> Nota ingresada: 7.5
> fin

Se registraron 4 notas.
Promedio: 7.75
```

````python
# ✍️ Escribe tu solución aquí:
entradas = ["8.0", "6.5", "9.0", "7.5", "fin"]
````

````python
# 👩‍🏫 SOLUCIÓN
entradas = ["8.0", "6.5", "9.0", "7.5", "fin"]
notas_acumuladas = []
i = 0

while True:
    entrada = entradas[i]           # en clase real: input("Ingresa una nota (o 'fin'): ")
    i += 1

    if entrada.lower() == "fin":
        print("> fin")
        break

    notas_acumuladas.append(float(entrada))
    print(f"> Nota ingresada: {entrada}")

if notas_acumuladas:
    promedio = sum(notas_acumuladas) / len(notas_acumuladas)
    print(f"\nSe registraron {len(notas_acumuladas)} notas.")
    print(f"Promedio: {promedio:.2f}")
else:
    print("No se ingresó ninguna nota.")
````

### 🟡 Ejercicio 4.2 — Conteo de calificaciones

Dado un grupo de notas, usa un diccionario para **contar cuántos estudiantes** cayeron en cada categoría:

```python
notas = [9.0, 5.5, 7.0, 4.0, 8.5, 6.0, 9.5, 3.0, 7.5, 8.0]
```

Categorías:
- `"Excelente"`: nota >= 9
- `"Bueno"`: nota >= 7
- `"Suficiente"`: nota >= 6
- `"Insuficiente"`: nota < 6

**Salida esperada:**
```
📊 Distribución del grupo:
  Excelente:    2 estudiantes
  Bueno:        3 estudiantes
  Suficiente:   1 estudiantes
  Insuficiente: 4 estudiantes
```

````python
# ✍️ Escribe tu solución aquí:
notas = [9.0, 5.5, 7.0, 4.0, 8.5, 6.0, 9.5, 3.0, 7.5, 8.0]
````

````python
# 👩‍🏫 SOLUCIÓN
notas = [9.0, 5.5, 7.0, 4.0, 8.5, 6.0, 9.5, 3.0, 7.5, 8.0]

conteo = {
    "Excelente": 0,
    "Bueno": 0,
    "Suficiente": 0,
    "Insuficiente": 0
}

for nota in notas:
    if nota >= 9:
        conteo["Excelente"] += 1
    elif nota >= 7:
        conteo["Bueno"] += 1
    elif nota >= 6:
        conteo["Suficiente"] += 1
    else:
        conteo["Insuficiente"] += 1

print("📊 Distribución del grupo:")
for categoria, cantidad in conteo.items():
    print(f"  {categoria:<13}: {cantidad} estudiantes")
````

salon = [

````python
# 👩‍🏫 SOLUCIÓN
salon = [
    {"nombre": "Lucía",   "notas": [8.0, 9.0, 7.5]},
    {"nombre": "Tomás",   "notas": [5.0, 4.5, 6.0]},
    {"nombre": "Valeria", "notas": [9.5, 8.5, 9.0]},
    {"nombre": "Andrés",  "notas": [6.0, 7.0, 5.5]},
    {"nombre": "Camila",  "notas": [7.5, 8.0, 8.5]},
]

print("╔══════════════════════════════════════╗")
print("║         BOLETÍN DEL GRUPO            ║")
print("╚══════════════════════════════════════╝")
print(f"{'NOMBRE':<12} {'PROMEDIO':>8}   ESTADO")
print("-" * 38)

promedios_grupo = []
aprobados = 0
reprobados = 0

for est in salon:
    nombre = est["nombre"]
    notas = est["notas"]
    promedio = sum(notas) / len(notas)
    promedios_grupo.append(promedio)

    if promedio >= 6.0:
        icono, estado = "✅", "Aprobado"
        aprobados += 1
    else:
        icono, estado = "❌", "Reprobado"
        reprobados += 1

    print(f"{icono} {nombre:<10} {promedio:>8.2f}   {estado}")

promedio_general = sum(promedios_grupo) / len(promedios_grupo)
print("-" * 38)
print(f"Promedio del grupo: {promedio_general:.2f}")
print(f"Aprobados: {aprobados}  |  Reprobados: {reprobados}")
````

horario = {
    "Lunes":     ["Matemáticas", "Español", "Tecnología"],
    "Martes":    ["Ciencias", "Educación Física", "Arte"],

    "Viernes":   ["Arte", "Matemáticas", "Educación Física"],
}
```

Escribe un programa que:
1. Imprima el horario completo, día por día
2. Cuente cuántas veces aparece `"Matemáticas"` en la semana
3. Encuentre qué día tiene más materias

**Salida esperada:**
```
📅 HORARIO SEMANAL
Lunes:
  1. Matemáticas
  2. Español
  3. Tecnología
...
Matemáticas aparece 3 veces en la semana.
El día con más materias es: Lunes (3 materias)

````python
# ✍️ Escribe tu solución aquí:
horario = {
    "Lunes":     ["Matemáticas", "Español", "Tecnología"],
    "Martes":    ["Ciencias", "Educación Física", "Arte"],
    "Miércoles": ["Matemáticas", "Historia", "Inglés"],
    "Jueves":    ["Tecnología", "Español", "Ciencias"],
    "Viernes":   ["Arte", "Matemáticas", "Educación Física"],
}
````

````python
# 👩‍🏫 SOLUCIÓN
horario = {
    "Lunes":     ["Matemáticas", "Español", "Tecnología"],
    "Martes":    ["Ciencias", "Educación Física", "Arte"],
    "Miércoles": ["Matemáticas", "Historia", "Inglés"],
    "Jueves":    ["Tecnología", "Español", "Ciencias"],
    "Viernes":   ["Arte", "Matemáticas", "Educación Física"],
}

# 1. Imprimir horario completo
print("📅 HORARIO SEMANAL")
print("=" * 30)
for dia, materias in horario.items():
    print(f"\n{dia}:")
    for i, materia in enumerate(materias, start=1):
        print(f"  {i}. {materia}")

# 2. Contar apariciones de Matemáticas
conteo_mat = 0
for dia, materias in horario.items():
    for materia in materias:
        if materia == "Matemáticas":
            conteo_mat += 1

print(f"\nMatemáticas aparece {conteo_mat} veces en la semana.")

# 3. Día con más materias
dia_max = ""
max_materias = 0
for dia, materias in horario.items():
    if len(materias) > max_materias:
        max_materias = len(materias)
        dia_max = dia

print(f"El día con más materias es: {dia_max} ({max_materias} materias)")
````
